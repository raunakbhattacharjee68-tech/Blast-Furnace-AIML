```python
temp_cols=['FUR_TOP_UPTAKE_TEMP1','FUR_TOP_UPTAKE_TEMP2','FUR_TOP_UPTAKE_TEMP3','FUR_TOP_UPTAKE_TEMP4']
df['Temp_Imbalance']=(df[temp_cols].max(axis=1)-df[temp_cols].min(axis=1))

press_cols=['FUR_UPTAKE_PRESS1','FUR_UPTAKE_PRESS2','FUR_UPTAKE_PRESS3','FUR_UPTAKE_PRESS4']
df['Pressure_Imbalance']=(df[press_cols].max(axis=1)-df[press_cols].min(axis=1))

df['Gas_Utilization']=(df['TOP_GAS_CO2_ANAL_TRX']/(df['TOP_GAS_CO_ANAL_TRX']+df['TOP_GAS_CO2_ANAL_TRX']))

df['Total_Oxygen']=(df['OXYGEN_FLOW_FLOW1']+df['OXYGEN_FLOW_FLOW2'])

df['Avg_HBT']=(df['HOT_BLAST_MAIN_TEMP1']+df['HOT_BLAST_MAIN_TEMP2'])/2

df['K_LM']=(df['LOWER_K']/df['MIDDLE_K'])
df['K_MU']=(df['MIDDLE_K']/df['UPPER_K'])
```

```python
features = [
    'Temp_Imbalance',
    'Pressure_Imbalance',
    'Gas_Utilization',
    'TOTAL_K',
    'K_LM',
    'K_MU'
]

# Select features
X_channel = (
    df[features]
    .replace([np.inf, -np.inf], np.nan)
    .dropna()
)

print(X_channel.shape)
```

```python
sample_idx = np.random.choice(
    len(X_channel),
    5000,
    replace=False
)

X_sample = X_channel.iloc[sample_idx]

print(X_sample.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X_sample)

print(X_scaled.shape)
```

```python
from sklearn.neighbors import NearestNeighbors

k = 12

neighbors = NearestNeighbors(n_neighbors=k)
neighbors_fit = neighbors.fit(X_scaled)

distances, indices = neighbors_fit.kneighbors(X_scaled)

distances = np.sort(distances[:, k-1])

plt.figure(figsize=(10,5))
plt.plot(distances)
plt.grid()
plt.show()
```

```python
from sklearn.cluster import DBSCAN

db = DBSCAN(
    eps=2.2,
    min_samples=12
)

labels = db.fit_predict(X_scaled)

print(pd.Series(labels).value_counts())
```

```python
from sklearn.decomposition import PCA

X_plot = PCA(n_components=2).fit_transform(X_scaled)

plt.figure(figsize=(10,8))

for k in set(labels):
    mask = labels == k

    plt.scatter(
        X_plot[mask,0],
        X_plot[mask,1],
        s=10,
        label=f'Cluster {k}'
    )

plt.legend()
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("DBSCAN Clusters")
plt.show()
```

```python
X_sample = X_sample.copy()
X_sample['cluster'] = labels

cluster_profile = X_sample.groupby('cluster').mean().round(2)

print(cluster_profile)
```

```python
normal = X_sample[X_sample['cluster']==0]

anoms = X_sample[X_sample['cluster']==-1]

print(normal.describe())
print(anoms.describe())
```

```python
for col in features:
    plt.figure(figsize=(8,4))

    sns.boxplot(
        data=X_sample,
        x='cluster',
        y=col
    )

    plt.title(col)
    plt.show()
```

```python
sample_df = df.loc[X_sample.index].copy()

sample_df['cluster'] = labels

print(sample_df['cluster'].value_counts())
```

```python
X_sample = X_sample.copy()

X_sample['cluster'] = labels

X_sample['Channeling_Flag'] = (
    X_sample['cluster'] == -1
).astype(int)
```

```python
comparison = X_sample.groupby(
    'Channeling_Flag'
)[features].mean()

print(comparison)
```

```python
comparison.T.plot(
    kind='bar',
    figsize=(10,5)
)

plt.title("Normal vs Potential Channeling")

plt.ylabel("Mean Value")

plt.xticks(rotation=45)

plt.show()
```

```python
channeling_rate = (
    X_sample['Channeling_Flag'].mean()
    * 100
)

print(
    f"Potential Channeling Rate: {channeling_rate:.2f}%"
)
```
